# Aplicando buenas practicas
1. python -m venv .venv                    (crear entorno aislado)
2. .venv\Scripts\activate                  (activar entorno aislado)
3. pip install pandas                      (comando para instalar cualquier libreria en python)
4. pip freeze > requeriments.txt           (comando para poner dependencias en el txt)
5. pip install -r requeriments.xtx         (comando para instalar dependencias)
6. Add-content .gitignore ".env" "venv\"   (comando para agregar hacia el .gitignore, aqui agregarmos nuestra carpeta venv y el .env)
